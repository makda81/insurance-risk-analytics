## Loss Ratio Analysis

Loss Ratio is an important insurance performance metric that measures the relationship between claims paid and premiums collected.

The formula used is:

\[
Loss\ Ratio = \frac{TotalClaims}{TotalPremium}
\]

The overall portfolio Loss Ratio is **0.528**, meaning that approximately **52.8% of collected premiums are used to cover claims**.

### Loss Ratio by Province

The loss ratio varies across provinces:

- Somali has the highest loss ratio (**0.612**), indicating that customers in this province generate relatively higher claims compared with premiums collected.
- Oromia has the second highest loss ratio (**0.537**).
- Amhara has the lowest loss ratio (**0.478**).

Although differences exist between provinces, the variation is moderate, which is consistent with the previous RiskScore analysis where provincial risk differences were relatively small.

### Loss Ratio by Vehicle Type

Vehicle type shows a much stronger difference in loss ratio:

- Luxury vehicles have the highest loss ratio (**0.842**), meaning claims are significantly higher compared with premiums collected.
- SUVs have a higher loss ratio (**0.564**) compared with standard vehicle categories.
- Hatchbacks (**0.420**) and Sedans (**0.404**) have lower loss ratios.

This supports the previous RiskScore analysis, where Luxury vehicles also showed the highest average risk score (**75.45**). Vehicle type appears to be an important factor influencing insurance risk and should be considered during feature engineering.

### Loss Ratio by Gender

Loss ratios are almost identical between genders:

- Female customers: **0.529**
- Male customers: **0.528**

This suggests that gender alone has limited impact on portfolio profitability compared with variables such as vehicle type.

### Business Insight

The analysis indicates that vehicle characteristics have a stronger relationship with insurance performance than demographic variables. Luxury vehicles represent a higher-risk segment due to their high claim-to-premium ratio, while province and gender show smaller differences.

## Claim Analysis by Vehicle Make

Average TotalClaims were compared across different vehicle manufacturers.

The results show clear differences between vehicle makes:

- Mercedes-Benz has the highest average claim amount (**3787.01**).
- BMW has the second highest average claim amount (**3362.89**).
- Lifan has the lowest average claim amount (**918.10**), followed by Hyundai (**1042.74**) and Suzuki (**1122.03**).

Higher claim amounts among premium vehicle brands may be related to higher repair costs, replacement values, and vehicle complexity.

These findings support the earlier RiskScore analysis, where Luxury vehicles showed significantly higher risk scores compared with other vehicle categories.

## Claim Analysis by Vehicle Model

Vehicle models also show differences in average claim amounts.

The highest average claim amounts were observed in:

- Land Cruiser (**3888.30**)
- E-Class (**3787.01**)
- 5 Series (**3362.89**)

The lowest average claim amounts were observed in:

- Elantra (**810.13**)
- Swift (**834.46**)
- 620 (**918.10**)

Premium vehicle models generally show higher claim amounts, suggesting that vehicle value and repair costs may contribute to claim severity.

These results indicate that vehicle-related features such as VehicleType, AutoMake, and VehicleModel may provide useful information for future machine learning models.
