```python
province_risk = (
    df.groupby("Province")
    ["RiskScore"]
    .mean()
    .sort_values()
)

province_risk
```

```python
province_risk.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Average Risk Score by Province")

plt.show()
```

```python
vehicle_risk = (
    df.groupby("VehicleType")
    ["RiskScore"]
    .mean()
)

vehicle_risk
```

```python
vehicle_risk.plot(
    kind="bar"
)

plt.title(
    "Risk Score by Vehicle Type"
)

plt.show()
```

# EDA Findings

1. The portfolio contains customers across multiple provinces and vehicle types.

2.The RiskScore distribution indicates that most customers fall within the medium-risk range, while fewer customers are classified at the extreme low and high ends. This suggests the portfolio is concentrated around moderate-risk customers.

3. Claim amounts show significant variation.The dataset is imbalanced, with approximately 15% of customers submitting claims.

4. ## Risk Score by Vehicle Type

The analysis shows that vehicle type is associated with differences in average risk scores.

Luxury vehicles have the highest average RiskScore (75.45), significantly higher than Hatchbacks (54.36) and Sedans (54.03).

SUVs also demonstrate elevated risk scores (60.57) compared with standard vehicle categories.

This suggests that vehicle type may be an important feature for predicting insurance risk during machine learning modeling.

5. ## Risk Score by Province

The analysis shows that average RiskScore values are relatively similar across provinces.

Somali has the highest average RiskScore (58.88), while Amhara has the lowest (57.80). However, the difference between provinces is small, suggesting that province alone may have limited influence on risk prediction.
